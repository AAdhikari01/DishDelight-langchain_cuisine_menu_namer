```python

#Prompt for restaurant name (one at a time)
prompt_name = PromptTemplate(
    input_variables=['cuisine'],
    template="Give me a fancy {cuisine} restaurant name."
)
name_chain = prompt_name | llm

#Prompt for menu
prompt_menu = PromptTemplate(
    input_variables=['restaurant_name'],
    template="Suggest me a concise menu (just dish names) for the restaurant called {restaurant_name}."
)
menu_chain = prompt_menu | llm

#Sequential function for one restaurant
def sequential_chain_single(cuisine):
    # Generate restaurant name
    restaurant_name = name_chain.invoke({"cuisine": cuisine}).content.strip()
    
    # Generate menu for that restaurant
    menu = menu_chain.invoke({"restaurant_name": restaurant_name}).content.strip()
    
    # Return in clean format
    return f"Restaurant Name: {restaurant_name}\nMenu: {menu}"

# Run it
output = sequential_chain_single("Nepali")
print(output)
```
